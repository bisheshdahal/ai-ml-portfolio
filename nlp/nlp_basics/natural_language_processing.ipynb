## Stemming -
Stemming is a text-preprocessing technique in computer science that cuts off word suffixes and prefixes to reduce words to their base or root form, such as "running" to "run". It helps search engines and text programs match similar words quickly.

```python
# Porter Stemmer
# Regexp Stemmer
# Snowball Stemer
```

```python
words = ['eating','eats','eaten','writing','writes','programming','programs','history','finally','finalized']
```

```python
from nltk.stem import PorterStemmer
stemming = PorterStemmer()
for word in words:
    print(word + '----->' + stemming.stem(word))
```

### POS important tags

- NN – noun, singular (dog)
- NNS – noun, plural (dogs)
- NNP – proper noun, singular (Nabin)
- NNPS – proper noun, plural (Americans)
- VB – verb, base form (run)
- VBD – verb, past tense (ran)
- VBG – verb, gerund (running)
- VBN – verb, past participle (eaten)
- VBP – verb, present, non-3rd (I run)
- VBZ – verb, present, 3rd sg (he runs)
- JJ – adjective (big)
- JJR – comparative adjective (bigger)
- JJS – superlative adjective (biggest)
- RB – adverb (quickly)
- RBR – comparative adverb (faster)
- RBS – superlative adverb (fastest)
- PRP – personal pronoun (I, you, he)
- PRP$ – possessive pronoun (my, your)
- DT – determiner (the, a)
- IN – preposition/conjunction (in, because)
- CC – coordinating conjunction (and, but)
- CD – cardinal number (one, 2026)
- TO – "to"
- MD – modal (can, will)
- UH – interjection (wow, hey)

```python
import nltk
from nltk.corpus import stopwords, wordnet
from nltk.tokenize import sent_tokenize , word_tokenize
from nltk.stem import WordNetLemmatizer

def get_wordnet_pos(treebank_tag):
    if treebank_tag.startswith('J'):
        return wordnet.ADJ
    elif treebank_tag.startswith('V'):
        return wordnet.VERB
    elif treebank_tag.startswith('N'):
        return wordnet.NOUN
    elif treebank_tag.startswith('R'):
        return wordnet.ADV
    else:
        return wordnet.NOUN
```

### BOW
Bag-of-Words (BoW) is a foundational feature extraction technique in Natural Language Processing (NLP) used to convert text data into numerical vectors so machine learning algorithms can process it

```python
'''
I love NLP because I love AI

Vocabulary :-----> I love NLP because AI

Words             Count
I                 2
Love              2
NLP               1
Because           1
AI                1

BOW Vector --> [2,2,1,1,1]
'''
```

TF-IDF, short for Term Frequency-Inverse Document Frequency, is a classic and widely used text vectorization technique in NLP. It builds directly upon the Bag-of-Words (BoW) model but introduces a crucial improvement: instead of just counting how many times a word appears, it assigns a weight to each word to reflect how important it is to a specific document within a larger collection (corpus).

```python
'''
TF(love) = 2/7
TF(NLP) = 1/7

'''
```

In Natural Language Processing (NLP), IDF stands for Inverse Document Frequency. It is the specific half of the TF-IDF algorithm that calculates how important or unique a word is across an entire collection of documents (a corpus).

```python
'''
Total 1000 Documents
the words appearing in 990

IDF = LOG(1000/990)
'''
import math
a= math.log(1000/990)
a*100
```
