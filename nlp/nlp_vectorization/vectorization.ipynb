```python
import pandas as pd
```

```python
messages = pd.read_csv('SMSSpamCollection', sep='\t', names=['label', 'message'])
messages
```

```python
import re
import nltk
```

```python
from nltk.corpus import stopwords
stopwords.words('english')
```

```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
ps.stem('running')
```

```python
stop_words = set(stopwords.words('english'))
# Builds the set of English stopwords ONCE, before the loop starts.

corpus = []
# Creates an empty list that will hold the cleaned version of each message.

for i in range(0, len(messages)): # Loops through every row/message in the 'messages' dataset by index.




    review = re.sub('[^a-zA-Z]', ' ', messages['message'][i])
    # Takes the i-th message text.
    # re.sub replaces everything that is NOT an alphabet letter (a-z, A-Z)
    # — so numbers, punctuation, emojis, special chars — with a space.
    # Example: "Hey! I won $500 🎉" -> "Hey  I won      "
    
    
    review = review.lower()
    # Converts the whole string to lowercase.


    review = review.split()
    # Splits the string into a list of individual words (tokenization),


    
    review = [ps.stem(word) for word in review if not word in stop_words]
    # This does two things in one list comprehension:
    # 1. Filters OUT stopwords (common words like "i", "the", "is", "and"
    #    that don't carry much meaning) using the pre-built stop_words set.
    # 2. For every remaining word, applies stemming (ps.stem) — reduces
    #    words to their root form.
    #    Example: "running" -> "run", "loved" -> "love"
    # Now using stop_words (built once above) instead of rebuilding
    # set(stopwords.words('english')) every iteration — this was the
    # main reason the loop was slow.



    review = ' '.join(review)
     # Joins the cleaned, stemmed list of words back into a single
    # space-separated string.

    corpus.append(review)
     # Adds this cleaned message string to the corpus list.
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=1000, binary=True)
X = cv.fit_transform(corpus).toarray()
X
```
